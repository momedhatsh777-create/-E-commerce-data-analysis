```python
customer_sales = orders.groupby("CustomerID")["TotalAmount"].sum()
customer_sales.sort_values(ascending=False).head(10)
```

```python
product_sales = orders_details.groupby("ProductID")["Quantity"].sum()
product_sales .sort_values(ascending=False).head(10)
```

```python
total_sales = orders["TotalAmount"].sum()
total_sales
```

```python
category_sales = (
orders_details
.merge(
    products[["productid", "categoryid"]],
    left_on="ProductID",
    right_on="productid"
)    
.merge
    (categories[["categoryid", "categoryname"]], 
    on="categoryid")
.groupby("categoryname")["Quantity"]
.sum()
.sort_values(ascending=False)
)
category_sales
```

```python
orders["OrderDate"] = pd.to_datetime(orders["OrderDate"])
monthly_sales =(
    orders
    .groupby(orders["OrderDate"].dt.to_period("m"))["TotalAmount"]
    .sum()
    .sort_index()
)
monthly_sales
```

```python
employee_orders = (
    orders
    .merge(
        employees[["EmployeeID", "EmployeeName"]],
        on="EmployeeID"
     )
 )
employee_orders.head()


    
```

```python
orders_per_employee = (
    employee_orders
    .groupby("EmployeeName")["OrderID"]
    .count()
    .reset_index(name="OrderCount")    
    .sort_values("OrderCount", ascending=False)
)
orders_per_employee
```

```python
top_employee = orders_per_employee.loc[
    orders_per_employee["OrderCount"].idxmax()
]
top_employee
```

```python
customer_orders = (
    orders
        .groupby("CustomerID")["OrderID"]
        .count()
        .sort_values(ascending=False)
)  
customer_orders
```

```python
customers.columns
```

```python
customer_orders_named = (
    customer_orders
    .reset_index()
    .merge(
        customers[["customerid","fullname"]],
        left_on="CustomerID",
        right_on="customerid"
    )
)
customer_orders_named
    
```

```python
customer_orders_named = customer_orders_named[
    ["fullname", "OrderID"]
]
customer_orders_named
```

```python

top_customers = customer_orders_named.head(10)

plt.figure(figsize=(10, 6))

plt.bar(
    top_customers["fullname"],
    top_customers["OrderID"]
)
plt.title("top 10 customers by number of orders")
plt.xlabel("customer")
plt.ylabel("number of orders")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
top_employees = orders_per_employee.head(10)

plt.figure(figsize=(10, 6))

plt.bar(
    top_employees.index.astype(str),
    top_employees.iloc[:, 0]
)
plt.title("top employees by number of orders")
plt.xlabel("Employee")
plt.ylabel("nimber of Orders")

plt.tight_layout()

plt.show()
```

```python
monthly_sales = orders.groupby(orders["OrderDate"].dt.month)["TotalAmount"].sum()
monthly_sales
```

```python
plt.figure(figsize=(10, 6))
plt.bar(
    monthly_sales.index.astype(str),
    monthly_sales.values
)
plt.title("monthly sales")
plt.xlabel("month")
plt.ylabel("total sales")

plt.tight_layout()
plt.show()
```
